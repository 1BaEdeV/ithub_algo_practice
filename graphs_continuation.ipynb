### Как читать код для графа

Граф состоит из **вершин** и **рёбер**. В социальной сети вершины — это люди, а ребро — дружба между двумя людьми.

- `nx.Graph()` создаёт обычный неориентированный граф: связь работает в обе стороны.
- `graph.add_node("Аня")` добавляет одну вершину.
- `graph.add_edge("Аня", "Боря")` соединяет две вершины.
- `graph.nodes` показывает вершины, а `graph.edges` — рёбра.
- `graph.neighbors("Аня")` показывает соседей Ани.
- `graph.degree("Аня")` показывает, сколько связей у Ани.

Сначала создаём объект `graph`, затем добавляем данные, а уже потом задаём вопросы графу.

````python
# Создаём маленькую сеть друзей и смотрим её свойства
import networkx as nx

friend_graph = nx.Graph()
friend_graph.add_node("Аня")
friend_graph.add_edge("Аня", "Боря")
friend_graph.add_edge("Аня", "Вера")
friend_graph.add_edge("Боря", "Глеб")

print("Вершины:", list(friend_graph.nodes))
print("Рёбра:", list(friend_graph.edges))
print("Соседи Ани:", list(friend_graph.neighbors("Аня")))
print("Степень Ани:", friend_graph.degree("Аня"))
````

### Обходы: BFS и DFS

**BFS** идёт по уровням: сначала проверяет ближайших соседей, потом соседей соседей. Это похоже на волну от сообщения в социальной сети.

**DFS** идёт в глубину: выбирает соседа и продолжает путь, пока это возможно, а потом возвращается назад. Это похоже на исследование одного маршрута до конца.

В обоих случаях важно множество `visited` — уже посещённые вершины. Без него граф с кругом друзей может заставить программу ходить по кругу.

````python
# Готовые BFS и DFS из networkx
friend_graph = nx.Graph()
friend_graph.add_edges_from([
    ("Аня", "Боря"),
    ("Аня", "Вера"),
    ("Боря", "Глеб"),
    ("Вера", "Даша"),
])

bfs_order = list(nx.bfs_tree(friend_graph, "Аня").nodes)
dfs_order = list(nx.dfs_preorder_nodes(friend_graph, "Аня"))

print("Порядок BFS:", bfs_order)
print("Порядок DFS:", dfs_order)
print("Первой в обоих обходах стоит:", bfs_order[0])
````

### Связность, кратчайшие пути и расписание

- `nx.connected_components(graph)` разбивает неориентированный граф на отдельные компании, где все могут добраться друг до друга.
- `nx.dijkstra_path(graph, start, finish, weight="weight")` ищет самый дешёвый путь, если у рёбер есть веса.
- `nx.dijkstra_path_length(...)` возвращает стоимость этого пути.
- `nx.topological_sort(graph)` работает с ориентированным графом зависимостей: например, сначала нужно выучить переменные, а потом функции.

Список, который возвращают некоторые функции `networkx`, удобно превращать в `list(...)`, чтобы его напечатать и увидеть обычные элементы Python.

````python
# Повторяем компоненты и кратчайший путь
road_graph = nx.Graph()
road_graph.add_weighted_edges_from([
    ("Дом", "Колледж", 3),
    ("Дом", "Магазин", 1),
    ("Магазин", "Колледж", 1),
    ("Парк", "Стадион", 2),
])

components = list(nx.connected_components(road_graph))
shortest_path = nx.dijkstra_path(road_graph, "Дом", "Колледж", weight="weight")
shortest_length = nx.dijkstra_path_length(road_graph, "Дом", "Колледж", weight="weight")

print("Компоненты:", components)
print("Самый дешёвый путь:", shortest_path)
print("Стоимость пути:", shortest_length)
````

````python
# Повторяем топологическую сортировку для расписания
schedule = nx.DiGraph()
schedule.add_edges_from([
    ("Переменные", "Условия"),
    ("Условия", "Циклы"),
    ("Циклы", "Функции"),
])

order = list(nx.topological_sort(schedule))
print("Порядок тем:", order)
````

## 2. Теория на простом языке

Цикл в графе — это как компания друзей, где можно пройти по связям и вернуться в исходную точку, не повторяя один и тот же путь в лоб. Если циклов нет, граф называют ацикличным. Такой граф удобно представлять как аккуратную сеть дорог без лишних кругов.

Минимальное остовное дерево — это способ оставить все вершины в одной сети, но с минимальной суммой весов рёбер. Представь, что ты соединяешь серверы интернетом или прокладываешь кабели между корпусами колледжа: хочется, чтобы всё работало, но без лишних затрат.

Двудольный граф делит вершины на две группы так, что рёбра идут только между группами, а не внутри них. В соцсетях это похоже на связь «пользователи — чаты» или «авторы — посты». А потоки в сетях помогают понять, сколько сообщений, заказов или данных реально пройдёт через сеть, если у каждого канала есть предел.

В этой паре ты увидишь, как всё это связано с обычными графами, которые ты уже умеешь строить, обходить и анализировать.

## 3. Разбор примера №1

````python
import networkx as nx
import matplotlib.pyplot as plt

# Собираем граф с кругом друзей
graph = nx.Graph()
graph.add_edges_from([
    ("Аня", "Боря"),
    ("Боря", "Вера"),
    ("Вера", "Аня"),
    ("Вера", "Глеб"),
    ("Глеб", "Даша"),
])

# Рисуем граф, чтобы увидеть цикл
plt.figure(figsize=(6, 4))
positions = nx.spring_layout(graph, seed=42)
nx.draw(graph, positions, with_labels=True, node_color="lightblue", node_size=1800, font_size=10, edge_color="gray")
plt.title("Круг друзей в социальной сети")
plt.show()

# Ищем циклы в графе
cycles = nx.cycle_basis(graph)
print("Найденные циклы:", cycles)
print("Количество циклов:", len(cycles))
````

## 4. Разбор примера №2

````python
# Смотрим на сеть серверов и ищем самый дешёвый каркас

graph = nx.Graph()
graph.add_weighted_edges_from([
    ("Сервер 1", "Сервер 2", 4),
    ("Сервер 1", "Сервер 3", 2),
    ("Сервер 2", "Сервер 3", 1),
    ("Сервер 2", "Сервер 4", 5),
    ("Сервер 3", "Сервер 4", 8),
    ("Сервер 3", "Сервер 5", 10),
    ("Сервер 4", "Сервер 5", 2),
    ("Сервер 2", "Сервер 5", 6),
])

plt.figure(figsize=(11, 4))
plt.subplot(1, 2, 1)
positions = nx.spring_layout(graph, seed=7)
nx.draw(graph, positions, with_labels=True, node_color="lightgreen", node_size=1700, font_size=9)
edge_labels = {}
for left, right, data in graph.edges(data=True):
    edge_labels[(left, right)] = data["weight"]
nx.draw_networkx_edge_labels(graph, positions, edge_labels=edge_labels, font_color="black")
plt.title("Исходная сеть")

plt.subplot(1, 2, 2)
mst = nx.minimum_spanning_tree(graph, weight="weight")
nx.draw(mst, positions, with_labels=True, node_color="orange", node_size=1700, font_size=9)
mst_labels = {}
mst_weight = 0
mst_edges = []
for left, right, data in mst.edges(data=True):
    mst_labels[(left, right)] = data["weight"]
    mst_weight = mst_weight + data["weight"]
    mst_edges.append((left, right, data["weight"]))
nx.draw_networkx_edge_labels(mst, positions, edge_labels=mst_labels, font_color="black")
plt.title("Минимальное остовное дерево")
plt.tight_layout()
plt.show()

print("Рёбра MST:", mst_edges)
print("Вес MST:", mst_weight)
````

## 16. Приложение

### Шпаргалка по функциям `networkx`

- `nx.Graph()` — неориентированный граф
- `nx.DiGraph()` — ориентированный граф
- `graph.add_node(...)`, `graph.add_edge(...)`, `graph.add_edges_from(...)` — добавить вершины и рёбра
- `graph.nodes`, `graph.edges`, `graph.degree` — посмотреть свойства графа
- `nx.cycle_basis(graph)` — найти базу циклов в неориентированном графе
- `nx.is_tree(graph)` — проверить, является ли граф деревом
- `nx.minimum_spanning_tree(graph, weight="weight")` — построить минимальное остовное дерево
- `nx.is_bipartite(graph)` — проверить двудольность
- `nx.maximum_flow(graph, source, sink, capacity="capacity")` — найти максимальный поток
- `nx.draw(...)`, `nx.draw_networkx_edge_labels(...)` — нарисовать граф и подписи к рёбрам

### Типичные ошибки

- **Перепутал вес и пропускную способность** — вес нужен для кратчайших путей и MST, а capacity — для потоков.
- **Ожидал двудольность, но в графе есть нечётный цикл** — такой граф уже не двудольный.
- **Думаешь, что MST всегда единственное** — на самом деле это не так.
- **Удалил не то ребро** — для дерева важно и связность, и отсутствие циклов одновременно.

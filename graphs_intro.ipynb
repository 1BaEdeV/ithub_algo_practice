## 5. Словари на графе

Словарь хранит пары «ключ — значение», как таблица с названием станции и количеством её соединений. Ключом будет вершина, а значением — её степень. Такой словарь помогает быстро узнать свойство конкретной вершины.

```python
import networkx as nx
graph = nx.Graph()
graph.add_edges_from([("A", "B"), ("A", "C"), ("B", "D")])
degree_table = {}
for node in graph.nodes:
    degree_table[node] = graph.degree[node]
print(degree_table)
print("Степень A:", degree_table["A"])
```

## 6. BFS — поиск в ширину

BFS идёт от стартовой вершины слоями: сначала проверяет ближайших друзей, затем друзей друзей. Это похоже на распространение новости по сети знакомств. Очередь `deque` помогает обрабатывать вершины в правильном порядке. В невзвешенном графе BFS находит кратчайший путь по числу рёбер.

```python
import networkx as nx
from collections import deque
graph = nx.Graph()
graph.add_edges_from([("A", "B"), ("A", "C"), ("B", "D"), ("C", "D")])
queue = deque(["A"])
visited = set(["A"])
order = []
while queue:
    node = queue.popleft()
    order.append(node)
    for neighbor in graph.neighbors(node):
        if neighbor not in visited:
            visited.add(neighbor)
            queue.append(neighbor)
print("Порядок BFS:", order)
print("Кратчайший путь:", nx.shortest_path(graph, "A", "D"))
```

## 7. DFS — поиск в глубину

DFS сначала идёт как можно дальше по одному направлению, а потом возвращается назад. Это похоже на исследование лабиринта. DFS можно написать рекурсивно, когда функция вызывает саму себя, или через стек. Стек обрабатывает последнюю добавленную вершину первой.

```python
import networkx as nx
graph = nx.Graph()
graph.add_edges_from([("A", "B"), ("A", "C"), ("B", "D")])
visited = set()
order = []
# def создаёт функцию
def dfs(node):
    visited.add(node)
    order.append(node)
    for neighbor in graph.neighbors(node):
        if neighbor not in visited:
            dfs(neighbor)
dfs("A")
print("DFS:", order)
```

## 9. Кратчайшие пути с весами: алгоритм Дейкстры

Иногда рёбра имеют вес: время поездки или длину дороги. Самый короткий путь тогда определяется общей суммой весов, а не количеством рёбер. Алгоритм Дейкстры выбирает лучший известный вариант. В `networkx` используются `nx.dijkstra_path` и `nx.dijkstra_path_length`.

```python
import networkx as nx
# Пример карты дорог с временем поездки.
graph = nx.Graph()
graph.add_edge("Дом", "Школа", weight=4)
graph.add_edge("Дом", "Парк", weight=1)
graph.add_edge("Парк", "Школа", weight=2)
graph.add_edge("Школа", "Магазин", weight=3)
path = nx.dijkstra_path(graph, "Дом", "Магазин", weight="weight")
length = nx.dijkstra_path_length(graph, "Дом", "Магазин", weight="weight")
print("Кратчайший путь:", path)
print("Время:", length)
```

## 10. Топологическая сортировка (DAG)

Ориентированный ациклический граф показывает зависимости между задачами. Например, сначала нужно купить продукты, а потом готовить ужин. Топологическая сортировка строит порядок, в котором все зависимости соблюдены. В таком графе нельзя вернуться по стрелкам в уже пройденную вершину.

```python
import networkx as nx
tasks = nx.DiGraph()
tasks.add_edge("Купить продукты", "Приготовить ужин")
tasks.add_edge("Приготовить ужин", "Поесть")
tasks.add_edge("Накрыть стол", "Поесть")
order = list(nx.topological_sort(tasks))
print("Порядок задач:", order)
```

```python
import networkx as nx
# Составь порядок подготовки к поездке.
tasks = nx.DiGraph()
tasks.add_edge("Собрать рюкзак", "Выйти из дома")
tasks.add_edge("Проверить билет", "Выйти из дома")
tasks.add_edge("Выйти из дома", "Сесть в поезд")
order = []
# Получи топологический порядок выполнения задач.
# YOUR CODE HERE
position = {}
for index in range(len(order)):
    position[order[index]] = index
assert position["Собрать рюкзак"] < position["Выйти из дома"], "Сначала собери рюкзак"
assert position["Проверить билет"] < position["Выйти из дома"], "Сначала проверь билет"
assert position["Выйти из дома"] < position["Сесть в поезд"], "Сначала выйди из дома"
print("Порядок подготовки:", " -> ".join(order))
print("✅ Верно!")
```
